```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("survey.csv")

print("Dataset loaded successfully!")

# First 5 rows
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\nDataset Shape:")
print(df.shape)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe(include="all"))
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

# Missing percentage
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

print("\nMissing Value Percentage:")
print(missing_percentage)
```

```python
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)
```

```python
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

df["Gender"].value_counts().head(10).plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Respondents")

plt.xticks(rotation=45)

plt.show()
```

```python

plt.figure(figsize=(6, 5))

sns.countplot(data=df, x="treatment")

plt.title("Mental Health Treatment")
plt.xlabel("Treatment")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="family_history",
    hue="treatment"
)

plt.title("Family History vs Mental Health Treatment")
plt.xlabel("Family History")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="work_interfere"
)

plt.title("Mental Health Work Interference")
plt.xlabel("Work Interference")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="benefits"
)

plt.title("Mental Health Benefits Provided by Employer")
plt.xlabel("Mental Health Benefits")
plt.ylabel("Number of Respondents")

plt.xticks(rotation=30)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="benefits",
    hue="treatment"
)

plt.title("Mental Health Benefits vs Treatment")
plt.xlabel("Mental Health Benefits")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="remote_work",
    hue="treatment"
)

plt.title("Remote Work vs Mental Health Treatment")
plt.xlabel("Remote Work")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="tech_company",
    hue="treatment"
)

plt.title("Tech Company vs Mental Health Treatment")
plt.xlabel("Tech Company")
plt.ylabel("Number of Respondents")

plt.show()
```

```python
country_counts = df["Country"].value_counts()

print("\nTop 10 Countries:")
print(country_counts.head(10))

plt.figure(figsize=(10, 6))

country_counts.head(10).plot(kind="bar")

plt.title("Top 10 Countries by Respondents")
plt.xlabel("Country")
plt.ylabel("Number of Respondents")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="supervisor"
)

plt.title("Willingness to Discuss Mental Health with Supervisor")
plt.xlabel("Response")
plt.ylabel("Number of Respondents")

plt.xticks(rotation=30)

plt.show()
```

```python
print("\nFamily History vs Treatment:")
print(
    pd.crosstab(
        df["family_history"],
        df["treatment"]
    )
)

print("\nBenefits vs Treatment:")
print(
    pd.crosstab(
        df["benefits"],
        df["treatment"]
    )
)

print("\nRemote Work vs Treatment:")
print(
    pd.crosstab(
        df["remote_work"],
        df["treatment"]
    )
)
```
